```python
import random

#0 = グー, 1 = チョキ, 2 = パー
hands = {
    0: "グー",
    1: "チョキ",
    2: "パー"
}

print("あなたの手を選択してください: グー, チョキ, パー")

player_input = input("あなたの手 (グー, チョキ, パー): ").strip()

player_hand_map = {
    "グー": 0,
    "チョキ": 1,
    "パー": 2
}

if player_input not in player_hand_map:
    print("無効な入力です。'グー', 'チョキ', 'パー' のいずれかを入力してください。")


player_hand = player_hand_map[player_input]

computer_choice_num = random.randint(0, 2)
computer_hand = hands[computer_choice_num]

print(f"あなた: {hands[player_hand]}")
print(f"コンピュータ: {computer_hand}")

result = (player_hand - computer_choice_num + 3) % 3

if result == 2:
    print("あなたの勝ちです")
elif result == 1:
    print("あなたの負けです")
else:
    print("引き分けです")

print("じゃんけんゲームを終了します。")
```
